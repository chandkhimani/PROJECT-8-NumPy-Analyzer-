```python
import numpy as np


class DataAnalytics:

    def __init__(self):
        self.array = None

    def set_array(self, array):
        self.array = np.array(array)

    def get_array(self):
        return self.array

    def display_array(self):
        if self.array is None:
            raise ValueError("Please create an array first.")

        print("\nCurrent Array:")
        print(self.array)

        print("\nArray Information")
        print("Dimension :", self.array.ndim)
        print("Shape     :", self.array.shape)
        print("Size      :", self.array.size)
        print("Data Type :", self.array.dtype)

    def get_element(self, index):
        if self.array is None:
            raise ValueError("Please create an array first.")

        return self.array[index]

    def slice_array(self, slicing):
        if self.array is None:
            raise ValueError("Please create an array first.")

        return self.array[slicing]

    def mathematical_operation(self, second_array, operation):

        if self.array is None:
            raise ValueError("Please create an array first.")

        if operation == "add":
            return self.array + second_array

        elif operation == "subtract":
            return self.array - second_array

        elif operation == "multiply":
            return self.array * second_array

        elif operation == "divide":

            if np.any(second_array == 0):
                raise ValueError("Division by zero is not allowed.")

            return self.array / second_array

        else:
            raise ValueError("Invalid operation.")

    def dot_product(self, second_array):

        if self.array is None:
            raise ValueError("Please create an array first.")

        return np.dot(self.array, second_array)

    def matrix_multiplication(self, second_array):

        if self.array is None:
            raise ValueError("Please create an array first.")

        if self.array.ndim != 2:
            raise ValueError(
                "Current array must be a 2D array."
            )

        return np.matmul(self.array, second_array)

    def combine_arrays(self, second_array, axis):

        if self.array is None:
            raise ValueError("Please create an array first.")

        return np.concatenate(
            (self.array, second_array),
            axis=axis
        )

    def split_array(self, sections, axis):

        if self.array is None:
            raise ValueError("Please create an array first.")

        return np.array_split(
            self.array,
            sections,
            axis=axis
        )

    def search_value(self, value):

        if self.array is None:
            raise ValueError("Please create an array first.")

        return np.where(self.array == value)

    def sort_array(self):

        if self.array is None:
            raise ValueError("Please create an array first.")

        return np.sort(self.array)

    def filter_array(self, condition, value):

        if self.array is None:
            raise ValueError("Please create an array first.")

        if condition == ">":
            return self.array[self.array > value]

        elif condition == "<":
            return self.array[self.array < value]

        elif condition == ">=":
            return self.array[self.array >= value]

        elif condition == "<=":
            return self.array[self.array <= value]

        elif condition == "==":
            return self.array[self.array == value]

        elif condition == "!=":
            return self.array[self.array != value]

        else:
            raise ValueError("Invalid condition.")

    def statistics(self):

        if self.array is None:
            raise ValueError("Please create an array first.")

        return {
            "Sum": np.sum(self.array),
            "Mean": np.mean(self.array),
            "Median": np.median(self.array),
            "Minimum": np.min(self.array),
            "Maximum": np.max(self.array),
            "Standard Deviation": np.std(self.array),
            "Variance": np.var(self.array)
        }


def convert_value(value):

    try:
        if "." in value:
            return float(value)
        else:
            return int(value)

    except ValueError:
        raise ValueError(
            f"{value} is not a valid number."
        )


def get_values():

    text = input(
        "Enter numbers separated by spaces: "
    ).strip()

    if not text:
        raise ValueError(
            "Please enter at least one number."
        )

    return [
        convert_value(x)
        for x in text.split()
    ]


def get_array_with_shape(shape):

    total = int(np.prod(shape))

    text = input(
        f"Enter {total} numbers separated by spaces: "
    ).strip()

    if not text:
        raise ValueError(
            "Please enter the required values."
        )

    values = [
        convert_value(x)
        for x in text.split()
    ]

    if len(values) != total:
        raise ValueError(
            f"Expected {total} values, "
            f"but received {len(values)}."
        )

    return np.array(values).reshape(shape)


def create_array(analyzer):

    print("\nCREATE NUMPY ARRAY")
    print("-" * 40)

    print("1. Create 1D Array")
    print("2. Create 2D Array")
    print("3. Create 3D Array")

    choice = input("Choose array type: ").strip()

    try:

        if choice == "1":

            values = get_values()

            analyzer.set_array(
                np.array(values)
            )

        elif choice == "2":

            rows = int(
                input("Enter number of rows: ")
            )

            columns = int(
                input("Enter number of columns: ")
            )

            if rows <= 0 or columns <= 0:
                raise ValueError(
                    "Rows and columns must be greater than zero."
                )

            array = get_array_with_shape(
                (rows, columns)
            )

            analyzer.set_array(array)

        elif choice == "3":

            depth = int(
                input("Enter depth: ")
            )

            rows = int(
                input("Enter number of rows: ")
            )

            columns = int(
                input("Enter number of columns: ")
            )

            if depth <= 0 or rows <= 0 or columns <= 0:
                raise ValueError(
                    "All dimensions must be greater than zero."
                )

            array = get_array_with_shape(
                (depth, rows, columns)
            )

            analyzer.set_array(array)

        else:

            print("Invalid choice.")
            return

        print("\nArray created successfully!")
        analyzer.display_array()

    except (ValueError, TypeError) as e:

        print("\nError:", e)


def indexing_slicing(analyzer):

    array = analyzer.get_array()

    if array is None:
        print("\nPlease create an array first.")
        return

    print("\nINDEXING AND SLICING")
    print("-" * 40)

    print("1. Get Element")
    print("2. Slice Array")

    choice = input("Enter your choice: ").strip()

    try:

        if choice == "1":

            if array.ndim == 1:

                index = int(
                    input("Enter index: ")
                )

                result = analyzer.get_element(index)

            elif array.ndim == 2:

                row = int(
                    input("Enter row index: ")
                )

                column = int(
                    input("Enter column index: ")
                )

                result = analyzer.get_element(
                    (row, column)
                )

            elif array.ndim == 3:

                depth = int(
                    input("Enter depth index: ")
                )

                row = int(
                    input("Enter row index: ")
                )

                column = int(
                    input("Enter column index: ")
                )

                result = analyzer.get_element(
                    (depth, row, column)
                )

            else:

                raise ValueError(
                    "Unsupported array dimension."
                )

            print("\nSelected Element:")
            print(result)

        elif choice == "2":

            if array.ndim == 1:

                start = int(
                    input("Enter start index: ")
                )

                end = int(
                    input("Enter end index: ")
                )

                result = analyzer.slice_array(
                    slice(start, end)
                )

            elif array.ndim == 2:

                row_start = int(
                    input("Enter row start: ")
                )

                row_end = int(
                    input("Enter row end: ")
                )

                column_start = int(
                    input("Enter column start: ")
                )

                column_end = int(
                    input("Enter column end: ")
                )

                result = analyzer.slice_array(
                    (
                        slice(row_start, row_end),
                        slice(column_start, column_end)
                    )
                )

            elif array.ndim == 3:

                depth_start = int(
                    input("Enter depth start: ")
                )

                depth_end = int(
                    input("Enter depth end: ")
                )

                row_start = int(
                    input("Enter row start: ")
                )

                row_end = int(
                    input("Enter row end: ")
                )

                column_start = int(
                    input("Enter column start: ")
                )

                column_end = int(
                    input("Enter column end: ")
                )

                result = analyzer.slice_array(
                    (
                        slice(depth_start, depth_end),
                        slice(row_start, row_end),
                        slice(column_start, column_end)
                    )
                )

            else:

                raise ValueError(
                    "Unsupported array dimension."
                )

            print("\nSliced Array:")
            print(result)

        else:

            print("Invalid choice.")

    except (ValueError, IndexError, TypeError) as e:

        print("\nError:", e)


def mathematical_operations(analyzer):

    if analyzer.get_array() is None:
        print("\nPlease create an array first.")
        return

    print("\nMATHEMATICAL OPERATIONS")
    print("-" * 40)

    print("1. Addition")
    print("2. Subtraction")
    print("3. Multiplication")
    print("4. Division")
    print("5. Dot Product")
    print("6. Matrix Multiplication")

    choice = input("Enter your choice: ").strip()

    try:

        if choice in ["1", "2", "3", "4"]:

            second = np.array(
                get_values()
            )

            operations = {
                "1": "add",
                "2": "subtract",
                "3": "multiply",
                "4": "divide"
            }

            result = analyzer.mathematical_operation(
                second,
                operations[choice]
            )

            print("\nResult:")
            print(result)

        elif choice in ["5", "6"]:

            rows = int(
                input("Enter rows of second array: ")
            )

            columns = int(
                input("Enter columns of second array: ")
            )

            second = get_array_with_shape(
                (rows, columns)
            )

            if choice == "5":

                result = analyzer.dot_product(
                    second
                )

            else:

                result = analyzer.matrix_multiplication(
                    second
                )

            print("\nResult:")
            print(result)

        else:

            print("Invalid choice.")

    except (ValueError, TypeError) as e:

        print("\nError:", e)


def combine_split(analyzer):

    array = analyzer.get_array()

    if array is None:
        print("\nPlease create an array first.")
        return

    print("\nCOMBINE / SPLIT ARRAYS")
    print("-" * 40)

    print("1. Combine Arrays")
    print("2. Split Array")

    choice = input("Enter your choice: ").strip()

    try:

        if choice == "1":

            print("\nCurrent Shape:")
            print(array.shape)

            axis = int(
                input(
                    f"Enter axis (0 to {array.ndim - 1}): "
                )
            )

            if axis < 0 or axis >= array.ndim:
                raise ValueError("Invalid axis.")

            shape = []

            for i in range(array.ndim):

                value = int(
                    input(
                        f"Enter size for dimension {i}: "
                    )
                )

                if value <= 0:
                    raise ValueError(
                        "Shape values must be greater than zero."
                    )

                shape.append(value)

            second = get_array_with_shape(
                tuple(shape)
            )

            result = analyzer.combine_arrays(
                second,
                axis
            )

            print("\nCombined Array:")
            print(result)

            print("\nNew Shape:")
            print(result.shape)

        elif choice == "2":

            sections = int(
                input("Enter number of sections: ")
            )

            axis = int(
                input(
                    f"Enter axis (0 to {array.ndim - 1}): "
                )
            )

            if axis < 0 or axis >= array.ndim:
                raise ValueError("Invalid axis.")

            result = analyzer.split_array(
                sections,
                axis
            )

            print("\nSplit Result:")

            for i, part in enumerate(
                result,
                start=1
            ):

                print(f"\nPart {i}:")
                print(part)

        else:

            print("Invalid choice.")

    except (ValueError, TypeError) as e:

        print("\nError:", e)


def search_sort_filter(analyzer):

    if analyzer.get_array() is None:
        print("\nPlease create an array first.")
        return

    print("\nSEARCH, SORT AND FILTER")
    print("-" * 40)

    print("1. Search Value")
    print("2. Sort Array")
    print("3. Filter Array")

    choice = input("Enter your choice: ").strip()

    try:

        if choice == "1":

            value = convert_value(
                input("Enter value to search: ")
            )

            result = analyzer.search_value(
                value
            )

            if all(len(x) == 0 for x in result):

                print("\nValue not found.")

            else:

                print("\nValue found at:")
                print(result)

        elif choice == "2":

            result = analyzer.sort_array()

            print("\nSorted Array:")
            print(result)

        elif choice == "3":

            print(
                "\nAvailable conditions:"
            )

            print(">  <  >=  <=  ==  !=")

            condition = input(
                "Enter condition: "
            ).strip()

            value = convert_value(
                input("Enter value: ")
            )

            result = analyzer.filter_array(
                condition,
                value
            )

            print("\nFiltered Array:")
            print(result)

        else:

            print("Invalid choice.")

    except (ValueError, TypeError) as e:

        print("\nError:", e)


def show_statistics(analyzer):

    try:

        result = analyzer.statistics()

        print("\nAGGREGATES AND STATISTICS")
        print("-" * 40)

        for key, value in result.items():

            print(
                f"{key:<22}: {value}"
            )

    except (ValueError, TypeError) as e:

        print("\nError:", e)


def main():

    analyzer = DataAnalytics()

    print("=" * 60)
    print("                 NUMPY ANALYZER")
    print("=" * 60)

    while True:

        print("\nMAIN MENU")
        print("-" * 40)

        print("1. Create a NumPy Array")
        print("2. Display Current Array")
        print("3. Indexing and Slicing")
        print("4. Perform Mathematical Operations")
        print("5. Combine or Split Arrays")
        print("6. Search, Sort and Filter")
        print("7. Aggregates and Statistics")
        print("8. Exit")

        print("-" * 40)

        choice = input(
            "Enter your choice: "
        ).strip()

        if choice == "1":

            create_array(analyzer)

        elif choice == "2":

            try:
                analyzer.display_array()

            except ValueError as e:
                print("\nError:", e)

        elif choice == "3":

            indexing_slicing(analyzer)

        elif choice == "4":

            mathematical_operations(analyzer)

        elif choice == "5":

            combine_split(analyzer)

        elif choice == "6":

            search_sort_filter(analyzer)

        elif choice == "7":

            show_statistics(analyzer)

        elif choice == "8":

            print("\nThank you for using NumPy Analyzer!")
            print("Program closed successfully.")

            break

        else:

            print(
                "\nInvalid choice. "
                "Please enter 1 to 8."
            )


main()
```
